```python
# Define incident generation settings

INCIDENT_PROBABILITY = 0.30

# Number of tickets that can belong to one incident
MIN_TICKETS_PER_INCIDENT = 2
MAX_TICKETS_PER_INCIDENT = 5

# Generate a unique incident ID
def generate_incident_id(incident_number):
    return f"INC{incident_number:05d}"

print("Incident probability:", INCIDENT_PROBABILITY)
print("Minimum tickets per incident:", MIN_TICKETS_PER_INCIDENT)
print("Maximum tickets per incident:", MAX_TICKETS_PER_INCIDENT)

print("Example incident ID:", generate_incident_id(1))
```

```python
# Define description details for each priority

priority_contexts = {
    "P1": [
        "A critical production service is completely unavailable.",
        "The outage is affecting multiple departments.",
        "A business-critical operation has stopped.",
        "The issue is causing a major service disruption."
    ],

    "P2": [
        "Several employees are unable to complete their work.",
        "An important business function is significantly affected.",
        "The issue is affecting an important internal service.",
        "The problem is disrupting normal department operations."
    ],

    "P3": [
        "The issue is affecting a limited number of employees.",
        "A normal business function is experiencing problems.",
        "The service is available but not working as expected.",
        "The issue has a workaround."
    ],

    "P4": [
        "The issue affects only one user.",
        "The request is related to a minor inconvenience.",
        "The user is requesting a non-urgent configuration change.",
        "There is no significant business impact."
    ]
}

# Verify priority descriptions

for priority, contexts in priority_contexts.items():
    print(f"\n{priority}:")
    for context in contexts:
        print("-", context)
```

```python
def generate_ticket_record(ticket_number, resolver_group, priority, client_id, incident_id):

    # Generate a unique ticket ID
    ticket_id = f"TKT{ticket_number:06d}"

    # Generate the ticket description
    description = generate_ticket_description(resolver_group)

    # Add context related to the assigned priority
    priority_context = random.choice(priority_contexts[priority])
    description += " " + priority_context

    # Generate a fictional caller name
    caller_name = f"User_{random.randint(1000, 9999)}"

    # Return the complete ticket
    return {
        "ticket_id": ticket_id,
        "text": description,
        "resolver_group": resolver_group,
        "priority": priority,
        "parent_incident_id": incident_id,
        "client_id": client_id,
        "caller_name": caller_name
    }

sample_ticket = generate_ticket_record(
    ticket_number=1,
    resolver_group="Network Operations",
    priority="P1",
    client_id="Client_A",
    incident_id="INC00001"
)

print(sample_ticket)
```

```python
# Total number of tickets
TOTAL_TICKETS = 10000

# Exact priority targets
priority_counts = {
    "P1": 500,
    "P2": 1500,
    "P3": 5000,
    "P4": 3000
}

# Store generated tickets
tickets = []

ticket_number = 1
incident_number = 1

# Generate tickets for each priority
for priority, target_count in priority_counts.items():

    generated_count = 0

    while generated_count < target_count:

        # Decide how many tickets belong to this incident
        if random.random() < INCIDENT_PROBABILITY:
            group_size = random.randint(
                MIN_TICKETS_PER_INCIDENT,
                MAX_TICKETS_PER_INCIDENT
            )
        else:
            group_size = 1

        # Do not exceed the target count
        group_size = min(
            group_size,
            target_count - generated_count
        )

        # Assign common incident properties
        resolver_group = random.choice(resolver_groups)
        client_id = random.choice(clients)

        incident_id = generate_incident_id(incident_number)
        incident_number += 1

        # Generate tickets belonging to this incident
        for _ in range(group_size):

            ticket = generate_ticket_record(
                ticket_number=ticket_number,
                resolver_group=resolver_group,
                priority=priority,
                client_id=client_id,
                incident_id=incident_id
            )

            tickets.append(ticket)

            ticket_number += 1
            generated_count += 1

# Convert records into a DataFrame
df = pd.DataFrame(tickets)

# Shuffle the dataset
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

print("Dataset generation completed.")
print("Total tickets:", len(df))
print("Total incidents:", df["parent_incident_id"].nunique())
print("Total resolver groups:", df["resolver_group"].nunique())
print("Total clients:", df["client_id"].nunique())
```
